```python
import deepchem as dc

# Define the featurizer parameters
acf = dc.feat.AtomicConvFeaturizer(
    frag1_num_atoms=100,     # Max atoms for ligand
    frag2_num_atoms=1000,    # Max atoms for protein pocket
    complex_num_atoms=1100,  # Sum of both
    max_num_neighbors=12,    # Max spatial neighbors per atom
    neighbor_cutoff=4.0      # Angstrom cutoff
)

# Load PDBBind using the complex featurizer
tasks, datasets, transformers = dc.molnet.load_pdbbind(
    featurizer=acf, 
    set_name='core',        # Use 'core' for quick testing, 'refined' for full training
    pocket=True             # Limits protein data to binding pocket to save memory
)

train_dataset, valid_dataset, test_dataset = datasets
```

```python
import deepchem as dc

# Define the grid featurizer
grid_feat = dc.feat.RdkitGridFeaturizer(
    voxel_width=16/15.0,
    feature_types=['ecfp', 'hbond', 'splif', 'salt_bridge', 'charge'],
)

# Load PDBBind
tasks, datasets, transformers = dc.molnet.load_pdbbind(
    featurizer=grid_feat, 
    set_name='core',

)
```

```python
# Build channel offsets for each per-voxel feature type in grid_feat.
X0 = np.nan_to_num(np.array(train.X, dtype=np.float16))[0]  # (nx, ny, nz, C)

# grid_feat.feature_types is a list of (is_flat, name) tuples.
voxel_types = [name for is_flat, name in grid_feat.feature_types if not is_flat]
flat_types  = [name for is_flat, name in grid_feat.feature_types if is_flat]
print("per-voxel:", voxel_types)
print("per-complex (not in X):", flat_types)

cuts = grid_feat.cutoffs
width_of = {
    "ecfp":        2 ** grid_feat.ecfp_power,
    "hbond":       len(cuts["hbond_dist_bins"]),
    "splif":       len(cuts["splif_contact_bins"]) * (2 ** grid_feat.splif_power),
    "salt_bridge": 1,
    "charge":      1,
    "sybyl":       len(grid_feat.sybyl_types),
    "pi_stack":    1,
    "cation_pi":   1,
}

offsets, start = {}, 0
for ft in voxel_types:
    w = width_of[ft]
    offsets[ft] = (start, start + w)
    start += w

print("computed total channels:", start, " actual X.shape[-1]:", X0.shape[-1])
print("offsets:", offsets)
```

```python
# Slice out just one feature type, e.g. hbond
ft = "hbond"
lo, hi = offsets[ft]
sub = X0[..., lo:hi]                     # shape (nx, ny, nz, width_of[ft])
print(ft, "sub-grid shape:", sub.shape)
print("nonzero voxels:", (sub.sum(axis=-1) > 0).sum())
print("per-channel totals:", sub.sum(axis=(0, 1, 2)))
```
